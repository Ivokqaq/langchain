```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_classic.agents import create_tool_calling_agent, AgentExecutor
from langchain_community.tools.tavily_search import TavilySearchResults
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

load_dotenv()

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

# Tavily - 搜索工具，用于联网查询（如天气数据） 需要 API key
search = TavilySearchResults(max_results=3, tavily_api_key=os.getenv("TAVILY_API_KEY"))

# Prompt：添加 MessagesPlaceholder 确保 agent_scratchpad 正确处理
prompt = ChatPromptTemplate.from_messages([
    ("system", "你是一个助手，必须用 tavily 工具联网查询。"),
    ("human", "{input}"),
    MessagesPlaceholder(variable_name="agent_scratchpad"),

# 创建 agent（绑定工具）
agent = create_tool_calling_agent(llm, [search], prompt)

# 使用 AgentExecutor
executor = AgentExecutor(
    agent=agent, 
    tools=[search], 
    verbose=True,
    handle_parsing_errors=True  # 处理工具调用解析失败
)

# 运行：注意输入格式
result = executor.invoke({"input": "今天北京天气"})
print(result["output"])
```

```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_tavily import TavilySearch
from langchain_experimental.tools import PythonREPLTool
from langchain import hub
from langchain.agents import create_react_agent, AgentExecutor

load_dotenv()

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

tools = [
    TavilySearch(max_results=3),
    PythonREPLTool(),
]

# 用官方现成的 ReAct prompt
prompt = hub.pull("hwchase17/react")

# 创建 agent
agent = create_react_agent(llm, tools, prompt)
executor = AgentExecutor(agent=agent, tools=tools, verbose=True, handle_parsing_errors=True)

# 提问
result = executor.invoke({"input": "特斯拉当前股价？比去年同期涨了多少百分比？"})
print("\n最终答案：")
print(result["output"])
```

```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage, SystemMessage
from langgraph.prebuilt import create_react_agent

load_dotenv()

@tool
def math_calculator(expr: str) -> str:
    """安全的数学计算器"""
    try:
        return str(eval(expr, {"__builtins__": {}}, {}))
    except:
        return "错误"

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

app = create_react_agent(llm, [math_calculator])
# 对比以前要写 8 行才能搞定的 AgentExecutor，现在一行就搞定
# 它内部自动帮你：
# - 绑定工具
# - 注入经典 ReAct prompt（hwchase17/react）
# - 创建完整的状态图（StateGraph）
# - 支持流式、记忆、断点续跑、多轮对话

system_prompt = """
你是一个绝对不能犯错的数学计算机器人，必须严格按照以下三步操作，缺一不可：

第一步：调用 math_calculator 计算 12345**2，得到准确的平方值
第二步：调用 math 工具把上一步结果 + 6789
第三步：调用 math 工具把上一步结果 / 5

每一步都必须单独调用工具，绝不能在脑子里算！
幂运算必须写成 **，绝不能写 ^ 或 pow()
开始！
"""

question = "请计算 12345 的平方，再加上 6789，然后除以 5，最后告诉我结果是多少？"

for chunk in app.stream(
    {
        "messages": [
            SystemMessage(content=system_prompt),
            HumanMessage(content=question)
        ]
    },
    stream_mode="values"
):
    msg = chunk["messages"][-1]
    msg.pretty_print()


print("\n=== 最终答案 ===")
result = app.invoke(
    {
        "messages": [
            SystemMessage(content=system_prompt),
            HumanMessage(content=question)
        ]
    }
)
print(result["messages"][-1].content)
```

```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_tavily import TavilySearch
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langgraph.prebuilt import create_react_agent

load_dotenv()

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
search = TavilySearch(max_results=2)

# 创建 Agent
app = create_react_agent(llm, [search])

# 手动管理记忆：用列表存储对话历史
conversation_history = []

# 第一轮：问北京
print("=== 第1轮：问北京 ===")
question1 = "北京明天天气怎么样？"

# 构建完整对话历史
messages1 = [
    SystemMessage(content="你是一个专业的天气助手，请使用搜索工具获取最新天气信息，用中文回答。"),
    HumanMessage(content=question1)
] + conversation_history

for chunk in app.stream(
    {"messages": messages1},
    stream_mode="values"
):
    msg = chunk["messages"][-1]
    if hasattr(msg, "pretty_print"):
        msg.pretty_print()

# 手动更新历史：添加 AI 回复
ai_response1 = chunk["messages"][-1].content
conversation_history.extend([
    HumanMessage(content=question1),
    AIMessage(content=ai_response1)
])

print(f"\nAI 回复1：{ai_response1}")

# 第二轮：问上海（会记住北京）
print("\n=== 第2轮：问上海（会记住北京） ===")
question2 = "那上海呢？"

# 构建完整对话历史（包含上一轮）
messages2 = [
    SystemMessage(content="你是一个专业的天气助手，请使用搜索工具获取最新天气信息，用中文回答。"),
    *conversation_history,  # 包含上一轮对话
    HumanMessage(content=question2)
]

for chunk in app.stream(
    {"messages": messages2},
    stream_mode="values"
):
    msg = chunk["messages"][-1]
    if hasattr(msg, "pretty_print"):
        msg.pretty_print()

# 手动更新历史
ai_response2 = chunk["messages"][-1].content
conversation_history.extend([
    HumanMessage(content=question2),
    AIMessage(content=ai_response2)
])

print(f"\nAI 回复2：{ai_response2}")

# 第三轮：测试记忆
print("\n=== 第3轮：测试记忆 ===")
question3 = "刚才两个城市哪个更冷？"

messages3 = [
    SystemMessage(content="你是一个专业的天气助手，请根据对话历史回答。"),
    *conversation_history,  # 包含前两轮完整历史
    HumanMessage(content=question3)
]

result3 = app.invoke({"messages": messages3})
print(f"\n记忆测试结果：{result3['messages'][-1].content}")

print(f"\n=== 完整对话历史 ===")
for i, msg in enumerate(conversation_history, 1):
    print(f"轮次 {i}: {msg.type} - {msg.content[:50]}...")
```

```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_tavily import TavilySearch
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langgraph.prebuilt import create_react_agent

load_dotenv()

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
search = TavilySearch(max_results=2)

# 创建 Agent
app = create_react_agent(llm, [search])

# 手动记忆列表
chat_history = []

# 永久系统提示
SYSTEM_PROMPT = "你是助手，可使用搜索工具。用中文清晰回答。"

def ask(question: str):
    """一键提问函数，自动带记忆"""
    global chat_history
    
    # 构建完整消息列表
    messages = [SystemMessage(content=SYSTEM_PROMPT)] + chat_history + [HumanMessage(content=question)]
    
    # 调用 Agent
    result = app.invoke({"messages": messages})
    
    # 提取 AI 回答
    ai_reply = result["messages"][-1].content
    
    # 更新记忆（只存用户和 AI 的消息）
    chat_history.extend([
        HumanMessage(content=question),
        AIMessage(content=ai_reply)
    ])
    
    return ai_reply

print(ask("北京明天天气"))
print(ask("上海呢？"))
print(ask("刚才两个城市哪个更冷？"))
```

```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_tavily import TavilySearch
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage
from langgraph.prebuilt import create_react_agent

load_dotenv()

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)
search = TavilySearch(max_results=2)

app = create_react_agent(llm, [search])

#手动系统提示
SYSTEM_PROMPT = "你是一个专业的中文天气助手，严格遵守以下规则：\n1. 用户问任何城市天气，必须调用 TavilySearch 工具\n2. 必须用中文清晰回答，包含城市名、日期、天气、最高/最低温度\n3. 永远记住自己是天气助手，不要问用户“请提供城市名”\n4. 当前日期是 2025年11月25日，明天是 11月26日"

chat_history = []  # 手动记忆

def ask(question: str):
    global chat_history
    
    # 每轮都强制把系统提示放在第一条
    messages = [SystemMessage(content=SYSTEM_PROMPT)] + chat_history + [HumanMessage(content=question)]
    
    result = app.invoke({"messages": messages})
    
    ai_reply = result["messages"][-1].content
    
    # 更新记忆
    chat_history.extend([HumanMessage(content=question), AIMessage(content=ai_reply)])
    
    return ai_reply

# 测试
print("第1轮：北京明天天气")
print(ask("北京明天天气"))
print("\n第2轮：上海呢？")
print(ask("上海呢？"))
print("\n第3轮：刚才两个城市哪个更冷？")
print(ask("刚才两个城市哪个更冷？"))
```
